
# Working with Jupyter Notebooks

Throughout this workshop, we will be working in **Jupyter Notebooks** and **JupyterLab**. These are industry-standard tools for data science and machine learning that allow you to combine executable code, rich text documentation, visualizations, and mathematical equations in a single, interactive document.

While standard Python scripts (`.py` files) are excellent for production code, Jupyter Notebooks (`.ipynb` files) excel at:

-   **Exploratory Data Analysis (EDA):** Seeing results immediately after running a block of code.
-   **Documentation:** Keeping notes alongside your logic.
-   **Visualizing:** Rendering plots and graphs directly within the workflow.

We will cover three ways to access this environment:

1.  **The Habrok HPC Cluster:** (Recommended for this course)
2.  **Your Local Machine:** (For working offline)
3.  **Google Colab:** (For quick prototyping)

# Using the Habrok HPC Cluster

For this course, we recommend using the university's High Performance Computing (HPC) cluster, **Habrok**. This ensures everyone has access to the same computational resources and hardware.

**NOTE**:

> You need to have a University of Groningen account to use Habrok. Your account also needs to be enable on Habrok, which you can request by following the [info](https://wiki.hpc.rug.nl/habrok/introduction/policies) on the wiki.
> 
> If you don't have a University account, you will have to use a local Jupyter installation, or Google Colab, as explained below.

## Accessing the Portal

1.  Navigate to the Habrok web portal: [https://portal.hb.hpc.rug.nl](https://portal.hb.hpc.rug.nl).
2.  Log in with your university credentials.
3.  We will use the **Jupyter Python Interactive App**, but first, we need to configure a custom environment to ensure all necessary libraries are installed.

## Building your own Jupyter Kernel on Habrok

A "Kernel" is the computational engine that executes the code contained in a Notebook document. By creating a custom kernel, you ensure your notebook uses the specific libraries required for this workshop.

### Step 1: Open a Shell Session

-   From the Habrok portal, go to the **"Clusters"** tab.
-   Select **"Habrok Login1 Shell"** (or similar) to open a terminal.

### Step 2: Load Python

We need to load a specific version of Python available on the cluster. Run the following command:

```shell
module load Python/3.13.1-GCCcore-14.2.0
```

> **Note:** If this specific version is unavailable in the future, you can search for available versions using `module spider Python`.

### Step 3: Create a Virtual Environment

It is best practice to keep project dependencies isolated. We will create a virtual environment (`venv`) in your home directory:

```shell
# Create the environment in a folder named 'venvs/ml_workshop'
python3 -m venv $HOME/venvs/ml_workshop
```

### Step 4: Activate the Environment

Before installing packages, you must activate this environment:

```shell
source $HOME/venvs/ml_workshop/bin/activate
```

*(Your terminal prompt should now show `(ml_workshop)` at the start.)*

### Step 5: Install Dependencies

Assuming you have the `requirements.txt` file in your current directory (if not, you may need to upload it or create it), install the packages:

```shell
pip install -r requirements.txt
```

### Step 6: Install the Jupyter Kernel

This step links your new virtual environment to the Jupyter interface so you can select it from a dropdown menu later.

```shell
python -m ipykernel install --user --name=ml_workshop --display-name="ML Workshop"
```

## Starting the Jupyter App

1.  Return to the **Habrok Web Portal**.
2.  Select the **Jupyter Python Interactive App**.
3.  **Configuration:**
    -   **Python Version:** Select `3.13.1` (matching the module you loaded).
    -   **Python Environment:** Select "Custom virtual environment".
    -   **Path:** Enter the path to your environment (e.g., `$HOME/venvs/ml_workshop`).
4.  Launch the app. Once inside the interface, you can ensure you are using the correct environment by checking the top-right corner of your notebook—it should say **"ML Workshop"**.

# Building your own Kernel on a Local Machine

Running Jupyter locally is great for offline work. However, the setup differs slightly depending on your operating system (Windows vs. macOS/Linux).

**Prerequisite:** Ensure Python is installed ([python.org/downloads](https://www.python.org/downloads/)).

## Step-by-Step Local Setup

### Open your terminal

-   **Windows:** Command Prompt or PowerShell.
-   **Mac/Linux:** Terminal.

### Create the Virtual Environment

Navigate to your project folder and run:

```shell
# Windows, Mac, and Linux
python -m venv venv
```

### Activate the Environment

This step differs by OS:

-   **Windows:**
    
    ```shell
    .\venv\Scripts\activate
    ```
-   **macOS / Linux:**
    
    ```shell
    source venv/bin/activate
    ```


### Install Dependencies

Install the course requirements plus the Jupyter Lab interface (which is not included in Python by default):

```shell
pip install -r requirements.txt
pip install jupyterlab ipykernel
```

### Register the Kernel

Just like on the cluster, we register this environment as a kernel:

```shell
python -m ipykernel install --user --name=ml_workshop_local --display-name="ML Workshop Local"
```

### Launch Jupyter Lab

Start the interface by running:

```shell
jupyter-lab
```

This will open a tab in your default web browser. When creating a new notebook, simply select **"ML Workshop Local"** from the Kernel menu.

# Working with Google Colab

Google Colab is a cloud-based notebook environment that requires no setup to start, though it has some limitations regarding session length and file storage. It is excellent for quick prototyping or if you are having trouble with local/HPC setups.

## Setup Instructions

1.  Log in to [Google Colab](https://colab.research.google.com/) with your Google Account.
2.  Create a **New Notebook**.

## Installing Requirements

Unlike local machines, Colab does not persist your environment after you close the tab. You must install libraries **every time you start a new session**.

### Option A: If you have a `requirements.txt` file

You need to upload the file to the Colab session first.

1.  Click the **Folder icon** (Files) on the left sidebar.
2.  Upload your `requirements.txt`.
3.  Run the following in a code cell:

````python
!pip install -r requirements.txt
````

### Option B: Direct Installation

If you only need a few specific packages, you can install them directly:

````python
!pip install pandas numpy scikit-learn
````

> **Pro Tip:** Google Colab offers free access to GPUs. To enable this, go to **Runtime > Change runtime type** and select **T4 GPU** under Hardware accelerator. This can significantly speed up training for deep learning models.
